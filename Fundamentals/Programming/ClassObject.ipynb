# Classes and Objects
Object-Oriented Programming like classes and objects helps structure and organized efficiently the code, to work with complex data and operations.

    - Combines data and operations
    - Manage complexity: Keep functions related together.
    - Improve reusability and organization (DSA)

## Class
Template or blueprint defines the structure and behavior of the object.

- Data members: Attributes
- Methods: Functions or operations

Benefits of using classes:
1. Describes the data and the functions to a particular entity.
2. Reusable template
3. Organize the code by grouping related variables and functions for an entity.
4. Modular programming: Easier to understand and maintain.

```python
class Car:
    def __init__(self, model="", year=""):
        self.model = model
        self.year = year
    
    def start_engine(self):
        return "Engine starting... \n"
    
    def stop_engine(self):
        return "Engine stoping... \n"
```

## Object
Instance of a class. Represents the real entity created using the template or blueprint (class)
- Stores values pre-defined on the class
- Availability to call methods 
- Multiple objects can exist from the same template



```python
myCar1 = Car()
myCar1.model = "Rav4"
myCar1.year = "2003"
print(f"Model: {myCar1.model} \nYear: {myCar1.year}")
print(myCar1.start_engine())
myCar1.stop_engine()
```

# Classes & Objects in Data Structures
Helps organize the data and the operations on that data. A class is the blueprint which defines the structure and behavior, while the object is the instance of that class.
- Reuse code: Write once, use as many times.
- Easier to represent real word entities: Stack, Queue and Linked-Lists

```python
class Stack:
    def __init__(self):
        self.arr = [0] * 5  # 5 value Stack
        self.top = -1 
    
    def push(self, x):
        if self.top == 4:
            print("Stack Overflow")
            return
        self.top += 1
        self.arr[self.top] = x

    def pop(self):
        if self.top == -1:
            print("Stack Underflow")
            return
        self.top -= 1

    def display(self):
        for i in range(self.top, -1, -1):
            if i == 0: print(self.arr[i])
            else: print(self.arr[i], end = "<-")
        print()
```

```python
s = Stack()

s.push(10)
s.push(20)
s.push(30)
s.push(40)
s.push(50)
s.push(60)  # 5th value -> StackOverflow
s.display()
```

```python
s.pop()
s.pop()
s.display()
```
