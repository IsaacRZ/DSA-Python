# Stack
Abstract data type which serves as a collection of elements. Two principal functions: push and pop (LIFO) Last-in First-out

    - push() -> Adds an element to the top of the stack
    - pop() ->  Removes the element of the top of the stack



```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class Stack:
    def __init__(self):
        self.top = None

    def push(self, value):
        new_node = Node(value)
        new_node.next = self.top
```
